```python
import math
from typing import Literal


def california_income_tax(
    taxable_income: float,
    filing_status: Literal["single", "married_joint", "married_separate", "head_of_household"] = "single",
) -> float:
 
    if not math.isfinite(taxable_income) or taxable_income < 0:
        raise ValueError("taxable_income must be finite and nonnegative.")

    single = (11_079, 26_264, 41_452, 57_542, 72_724, 371_479, 445_771, 742_953)
    thresholds = {
        "single": single,
        "married_separate": single,
        "married_joint": tuple(2 * threshold for threshold in single),
        "head_of_household": (
            22_173, 52_530, 67_716, 83_805, 98_990, 505_208, 606_251, 1_010_417
        ),
    }
    if filing_status not in thresholds:
        raise ValueError(f"Unsupported filing status: {filing_status!r}")

    rates = (0.01, 0.02, 0.04, 0.06, 0.08, 0.093, 0.103, 0.113, 0.123)
    upper_limits = (*thresholds[filing_status], math.inf)

    tax = 0.0
    lower = 0.0
    for upper, rate in zip(upper_limits, rates):
        tax += max(0.0, min(taxable_income, upper) - lower) * rate
        lower = upper
        if taxable_income <= upper:
            break

    tax += 0.01 * max(0.0, taxable_income - 1_000_000)
    return round(tax, 2)
```

```python
wealth = 100_000
compensation = 200_000

for x in range(1, 20+1):
    wealth *= 1.08
    compensation *= 1.03

    taxes = california_income_tax(compensation, filing_status="single")

    income = compensation - taxes
    saved = 0.2 * income
    wealth += saved

    print(
        f"Year {x:>2}:  "
        f"Compensation = ${compensation:>14,.2f}    "
        f"Income = ${income:>14,.2f}    "
        f"Income Taxes = ${taxes:>14,.2f}    "
        f"Saved = ${saved:>14,.2f}    "
        f"Wealth = ${wealth:>16,.2f}"
    )
```
